```python
from roboflow import Roboflow
from google.colab import userdata, files
import os
from ultralytics import YOLO
```

```python
ROBOFLOW_API_KEY = userdata.get('ROBOFLOW_API_KEY')
```

```python
# Initialize Roboflow
rf = Roboflow(api_key=ROBOFLOW_API_KEY)
```

```python
# Access the project and download the specific YOLOv8 version
project = rf.workspace("project-ida").project("structural-defects-cmies")
dataset = project.version(4).download("yolov8")
```

```python
# Initialize lightweight baseline YOLOv8 model
model = YOLO("yolov8n.pt")
```

```python
dataset_yaml = "/content/Structural-Defects-4/data.yaml"
```

```python
results = model.train(
    data=dataset_yaml,
    epochs=30,
    imgsz=640,
    device=0,
    project="structural_defects",
    name="yolov8n_base"
)
```
